```python
%%bash
mkdir -p turner_sigs

aria2c \
    --input=turner_accessions.txt \
    --dir turner_sigs/ \
    --continue \
    --timeout=10 \
    --connect-timeout=5 \
    --max-tries=2 \
    --retry-wait=3 \
    --max-connection-per-server=4

sourmash compare \
    ./turner_sigs/*.sig \
    --output turner_compare.np \
    --csv turner_compare.csv \
    --labels-to turner_compare_labels.csv \
    -k 51

sourmash plot \
    --labels-from turner_compare_labels.csv \
    turner_compare.np \
    --output turner_compare.png
```
