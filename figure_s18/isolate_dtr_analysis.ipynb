```python
import polars as pl

metadata = pl.read_csv('/mmfs1/gscratch/pedslabs_hoffman/carsonjm/CFPhageome/repos/uhvdb-manuscript/figure_s18/uhvdb_v5_final_metadata_v2.tsv.gz', separator='\t')
```

```python
print("Number of unique isolate viruses:",
    metadata
        .filter(pl.col('seq_name').str.contains('RSGB'))
        .filter(pl.col('source_db') == 'MOTUS_DB')
        .n_unique('hash')
)

print("Number of unique isolate DTR viruses:",
    metadata
        .filter(pl.col('seq_name').str.contains('RSGB'))
        .filter(pl.col('source_db') == 'MOTUS_DB')
        .filter((pl.col('topology') == 'DTR') | (pl.col('completeness_method').str.contains('DTR')))
        .n_unique('hash')
)
```

```python
(
    metadata
        .filter(pl.col('seq_name').str.contains('RSGB'))
        .filter(pl.col('source_db') == 'MOTUS_DB')
        .filter((pl.col('topology') == 'DTR') | (pl.col('completeness_method').str.contains('DTR')))
        .unique('hash')
        [['temperate', 'phrog_integration_excision', 'empathi_integration', 'integration_status']]
        .filter(pl.col('empathi_integration') >= 1)
)
```
