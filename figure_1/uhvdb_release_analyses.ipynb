```python
### Determine proportion unannotated
### Annotate uhvdb_r1_batka products from Bakta DB (UniRef90/100) + UniRef50-virus FAA (UniRef30)

import gzip
import sqlite3

BAKTA_DB = '/mmfs1/gscratch/pedslabs_hoffman/carsonjm/CFPhageome/repos/UHVDB/toolkit/databases/bakta/1.12_full/db/bakta.db'
UNIREF50_VIRUS_FAA = '/mmfs1/gscratch/pedslabs_hoffman/carsonjm/CFPhageome/repos/UHVDB/toolkit/databases/uniref50virus/2026_03/uniref50_virus.faa.gz'


def _is_unannotated_product(product):
    """True if product is missing or is a generic/hypothetical label."""
    if product is None:
        return True
    p = str(product).strip()
    if not p or p == '-':
        return True
    plower = p.lower()
    if plower in {'hypothetical protein', 'phage protein', 'putative protein'}:
        return True
    if 'hypothetical' in plower or plower.startswith('uncharacterized'):
        return True
    return False


# Parse UniRef level + ID from Bakta accessions
uhvdb_r1_batka = (
    uhvdb_r1_batka
    .with_columns([
        pl.col('Accession').str.extract(r'^UniRef:UniRef(100|90|30)_(.+)$', 1).alias('uniref_level'),
        pl.col('Accession').str.extract(r'^UniRef:UniRef(100|90|30)_(.+)$', 2).alias('uniref_id'),
    ])
)

ids_100 = set(uhvdb_r1_batka.filter(pl.col('uniref_level') == '100')['uniref_id'].drop_nulls().to_list())
ids_90 = set(uhvdb_r1_batka.filter(pl.col('uniref_level') == '90')['uniref_id'].drop_nulls().to_list())
ids_30 = set(uhvdb_r1_batka.filter(pl.col('uniref_level') == '30')['uniref_id'].drop_nulls().to_list())
print(f'Unique UniRef100 IDs: {len(ids_100):,}')
print(f'Unique UniRef90 IDs: {len(ids_90):,}')
print(f'Unique UniRef30 IDs: {len(ids_30):,}')

# --- UniRef30 products from expert UniRef50-virus FASTA headers ---
# Header: >UniRef30_XXX 30~~~80~~~80~~~RepID~~~product~~~UniRef:UniRef30_XXX
product_30 = {}
with gzip.open(UNIREF50_VIRUS_FAA, 'rt') as f_in:
    for line in f_in:
        if not line.startswith('>'):
            continue
        fields = line[1:].strip().split('~~~')
        if len(fields) < 5:
            continue
        uid = fields[0].split()[0].replace('UniRef30_', '').replace('UniRef50_', '')
        if uid in ids_30:
            product_30[uid] = fields[4]
print(f'UniRef30 IDs with product lookup: {len(product_30):,} / {len(ids_30):,}')

# --- UniRef90/100 products from bakta.db ---
con = sqlite3.connect(':memory:')
con.execute(f"ATTACH DATABASE 'file:{BAKTA_DB}?mode=ro' AS bakta")
con.execute('CREATE TABLE q100 (id TEXT PRIMARY KEY)')
con.execute('CREATE TABLE q90 (id TEXT PRIMARY KEY)')
con.executemany('INSERT OR IGNORE INTO q100(id) VALUES (?)', ((i,) for i in ids_100))
con.executemany('INSERT OR IGNORE INTO q90(id) VALUES (?)', ((i,) for i in ids_90))

map_100 = {
    u100: (u90, prod)
    for u100, u90, prod in con.execute(
        '''
        SELECT i.uniref100_id, i.uniref90_id, i.product
        FROM bakta.ips i
        JOIN q100 q ON q.id = i.uniref100_id
        '''
    )
}
print(f'UniRef100 IDs mapped via ips: {len(map_100):,} / {len(ids_100):,}')

needed_90 = set(ids_90) | {u90 for u90, _ in map_100.values() if u90}
con.execute('CREATE TABLE q90_all (id TEXT PRIMARY KEY)')
con.executemany('INSERT OR IGNORE INTO q90_all(id) VALUES (?)', ((i,) for i in needed_90))

product_90 = {
    u90: prod
    for u90, prod in con.execute(
        '''
        SELECT p.uniref90_id, p.product
        FROM bakta.psc p
        JOIN q90_all q ON q.id = p.uniref90_id
        '''
    )
}

missing_90 = {u for u, prod in product_90.items() if prod is None} | (needed_90 - set(product_90))
if missing_90:
    con.execute('CREATE TABLE q90_missing (id TEXT PRIMARY KEY)')
    con.executemany('INSERT OR IGNORE INTO q90_missing(id) VALUES (?)', ((i,) for i in missing_90))
    for u90, prod in con.execute(
        '''
        SELECT p.uniref90_id, c.product
        FROM bakta.psc p
        JOIN q90_missing q ON q.id = p.uniref90_id
        JOIN bakta.pscc c ON c.uniref50_id = p.uniref50_id
        WHERE c.product IS NOT NULL
        '''
    ):
        if product_90.get(u90) is None:
            product_90[u90] = prod
con.close()
print(f'UniRef90 IDs with non-null product: {sum(1 for v in product_90.values() if v is not None):,} / {len(needed_90):,}')

# Join product lookups onto Bakta hits
annot_30 = (
    pl.DataFrame({'uniref_id': list(product_30.keys()), 'product_30': list(product_30.values())})
    if product_30 else pl.DataFrame(schema={'uniref_id': pl.Utf8, 'product_30': pl.Utf8})
)
annot_90 = (
    pl.DataFrame({'uniref_id': list(product_90.keys()), 'product_90': list(product_90.values())})
    if product_90 else pl.DataFrame(schema={'uniref_id': pl.Utf8, 'product_90': pl.Utf8})
)
annot_100 = pl.DataFrame({
    'uniref_id': list(map_100.keys()),
    'product_100': [ips_prod or product_90.get(u90) for u90, ips_prod in map_100.values()],
}) if map_100 else pl.DataFrame(schema={'uniref_id': pl.Utf8, 'product_100': pl.Utf8})

uhvdb_r1_batka = (
    uhvdb_r1_batka
    .join(annot_30, on='uniref_id', how='left')
    .join(annot_90, on='uniref_id', how='left')
    .join(annot_100, on='uniref_id', how='left')
    .with_columns(
        pl.when(pl.col('uniref_level') == '30').then(pl.col('product_30'))
         .when(pl.col('uniref_level') == '90').then(pl.col('product_90'))
         .when(pl.col('uniref_level') == '100').then(pl.col('product_100'))
         .otherwise(None)
         .alias('product')
    )
    .drop(['product_30', 'product_90', 'product_100'])
    .with_columns(
        pl.col('product').map_elements(_is_unannotated_product, return_dtype=pl.Boolean).alias('unannotated')
    )
)

n_genes = uhvdb_r1_batka.n_unique('Locus Tag')
n_with_uniref = uhvdb_r1_batka.filter(pl.col('uniref_id').is_not_null()).n_unique('Locus Tag')
n_functional = uhvdb_r1_batka.filter(pl.col('unannotated') == False).n_unique('Locus Tag')
n_unannotated = uhvdb_r1_batka.filter(
    (pl.col('uniref_id').is_null()) | (pl.col('unannotated') == True)
).n_unique('Locus Tag')

print(f'Total Bakta genes (unique Locus Tag): {n_genes:,}')
print(f'With UniRef accession: {n_with_uniref:,} ({n_with_uniref / n_genes:.3%})')
print(f'With functional product annotation: {n_functional:,} ({n_functional / n_genes:.3%})')
print(f'Unannotated (no UniRef or generic/null product): {n_unannotated:,} ({n_unannotated / n_genes:.3%})')

uni = uhvdb_r1_batka.filter(pl.col('uniref_id').is_not_null())
n_uni = uni.n_unique('Locus Tag')
n_uni_unannot = uni.filter(pl.col('unannotated') == True).n_unique('Locus Tag')
print(f'Among UniRef hits, unannotated product: {n_uni_unannot:,} / {n_uni:,} ({n_uni_unannot / n_uni:.3%})')

uhvdb_r1_batka.select(['Locus Tag', 'Accession', 'uniref_level', 'uniref_id', 'product', 'unannotated']).head(10)
```
