```python
%%bash
### Create input file of fastas
ls /gscratch/scrubbed/carsonjm/2026.03.02/40/4c6c909e0cf1bc3c29ecf3ed759afc/host_fastas/* > fasta_files.txt

# create tsv with file base name for 1st column then path for second column
awk -F/ '{print $NF"\t"$0}' fasta_files.txt > fasta_files.tsv
sed -i 's/\.fa.*\t/\t/g' fasta_files.tsv
```

```python
%%bash
### Create sketchlib sketch
sketchlib sketch \
    -f fasta_files.tsv \
    -o chunk_102_s1k \
    -k 21 \
    -s 1000 \
    -v \
    --threads 16
# ~ 2 min
# ~ 200 min for all 2M
```

```python
%%bash
### Create sketchlib sketch
sketchlib sketch \
    -f fasta_files.tsv \
    -o chunk_102_1k \
    -k 21 \
    -s 1000 \
    -v \
    --threads 16
# ~ 4 min
# ~ 400 min for all 2M
```

```python
%%bash
### Create an inverted sketch for preclustering
sketchlib inverted build \
    -f fasta_files.tsv \
    --species-names /mmfs1/gscratch/pedslabs_hoffman/carsonjm/CFPhageome/repos/uhvdb-manuscript/host_genome_id/combined_inverted_clusters.tsv \
    -o chunk_102 \
    --write-skq \
    -s 100 \
    -k 21 \
    -v \
    --threads 32
# ~ 2 min for -s 10
# ~ 4 min for -s 100
# ~400 min for all 2M
```

```python
%%bash
### Calculate distance within sketch
sketchlib inverted precluster \
    chunk_102.ski \
    --skd chunk_102 \
    -o chunk_102_precluster \
    --ani \
    --threads 16 \
    --verbose \
    --knn 100
# ~10 s; 4 min total
```

```python
%%bash
### Calculate distance within sketch
sketchlib dist \
    chunk_102.skm \
    -o chunk_102_dist \
    -k 21 \
    --ani \
    --threads 16 \
    --verbose \
    --knn 100
# ~ 45 s for 15K
# ~ 500 s for 50K (3^2)
# ~ 6.8 hr for 350K (7^2) (potentially 4 hrs with 32 CPUs)
```

```python
%%bash
### Calculate distance within sketch
sketchlib dist \
    chunk_102_1k.skm \
    -o chunk_102_1k_dist \
    -k 21 \
    --ani \
    --threads 16 \
    --verbose \
    --knn 100
# ~ 6s for 15K
# ~ 53 s for 50K (3^2=8)
# ~ 1 hr for 350K (7^2) (potentially 1/2 hr with 32 CPUs)
```

```python
%%bash
### Calculate distance with skani
skani triangle \
    -l fasta_files.txt \
    -o skani_chunk_102 \
    --short-header \
    --sparse \
    --fast \
    -t 16 \
    -s 99.5
# ~10 min with default settings
```

```python
### Load data
import seaborn as sns
import matplotlib.pyplot as plt
import polars as pl

sketchlib_pre = (
    pl.read_csv('chunk_102_precluster', separator='\t', has_header=False, new_columns=['genome1', 'genome2', 'ani_prop'])
        .with_columns([
            (pl.col('ani_prop') * 100).alias('ani')
        ])
)

sketchlib_dist = (
    pl.read_csv('chunk_102_dist', separator='\t', has_header=False, new_columns=['genome1', 'genome2', 'ani_prop'])
        .with_columns([
            (pl.col('ani_prop') * 100).alias('ani')
        ])
)

sketchlib_1k_dist = (
    pl.read_csv('chunk_102_1k_dist', separator='\t', has_header=False, new_columns=['genome1', 'genome2', 'ani_prop'])
        .with_columns([
            (pl.col('ani_prop') * 100).alias('ani')
        ])
)

skani_data = (
    pl.read_csv('skani_chunk_102', separator='\t', columns=['Ref_name', 'Query_name', 'ANI', 'Align_fraction_ref', 'Align_fraction_query'])
        .rename({'ANI': 'ani', 'Ref_name': 'genome1', 'Query_name': 'genome2', 'Align_fraction_ref': 'ref_af', 'Align_fraction_query': 'q_af'})
        .with_columns([
            pl.col('genome1').str.replace(r'-scaffold.*', ''),
            pl.col('genome2').str.replace(r'-scaffold.*', ''),
            (pl.col('ani') * (pl.col('q_af')/100) * (pl.col('ref_af')/100)).alias('global_ani')
        ])
)
```

```python
### Compare skani and sketchlib ANI values
joined1 = (
    sketchlib_dist
        .join(skani_data, on=['genome1', 'genome2'], how='full', coalesce=True, suffix='_skani')
        .fill_null(0)
)
print("sketchlib >= 99.9% ANI:", joined1.filter(pl.col('ani') >= 99.9).shape[0])
print("skani >= 99.9% ANI:", joined1.filter(pl.col('global_ani') >= 99.9).shape[0])
# print("skani >= 99.9% ANI:", joined.filter((pl.col('ani_skani') >= 99.9) & (pl.col('ani') == 0)).shape[0])
```

```python
### Compare skani and sketchlib ANI values
joined2 = (
    sketchlib_pre
        .join(skani_data, on=['genome1', 'genome2'], how='full', coalesce=True, suffix='_skani')
        .fill_null(0)
)
print("sketchlib >= 99.9% ANI:", joined2.filter(pl.col('ani') >= 99.9).shape[0])
print("skani >= 99.9% ANI:", joined2.filter(pl.col('global_ani') >= 99.9).shape[0])
# print("skani >= 99.9% ANI:", joined.filter((pl.col('ani_skani') >= 99.9) & (pl.col('ani') == 0)).shape[0])
```

```python
### Compare skani and sketchlib ANI values
joined3 = (
    sketchlib_1k_dist
        .join(skani_data, on=['genome1', 'genome2'], how='full', coalesce=True, suffix='_skani')
        .fill_null(0)
)
print("sketchlib >= 99.9% ANI:", joined3.filter(pl.col('ani') >= 99.9).shape[0])
print("skani >= 99.9% ANI:", joined3.filter(pl.col('global_ani') >= 99.9).shape[0])
# print("skani >= 99.9% ANI:", joined.filter((pl.col('ani_skani') >= 99.9) & (pl.col('ani') == 0)).shape[0])
```
