```python
%%bash
### Create input file of fastas
ls /gscratch/scrubbed/carsonjm/2026.03.02/40/4c6c909e0cf1bc3c29ecf3ed759afc/host_fastas/* > fasta_files.txt
ls /gscratch/scrubbed/carsonjm/2026.03.02/f4/c8e78f40f28c8a356a7a80e26b1308/host_fastas/* >> fasta_files.txt
ls /gscratch/scrubbed/carsonjm/2026.03.02/3f/cc37cb99578e05f7817f23da828f4e/host_fastas/* >> fasta_files.txt

# create tsv with file base name for 1st column then path for second column
awk -F/ '{print $NF"\t"$0}' fasta_files.txt > fasta_files.tsv
sed -i 's/\.fa.*\t/\t/g' fasta_files.tsv
```

```python
%%bash
### Create sketchlib sketch
sketchlib sketch \
    -f fasta_files.tsv \
    -o chunk_102 \
    -k 21 \
    -s 10000 \
    -v \
    --threads 16
# Takes about 8min for 50,000 genomes
# Size = 922 MB
```

```python
%%bash
### Create sketchlib sketch
sketchlib sketch \
    -f fasta_files.tsv \
    -o chunk_102_1k \
    -k 21 \
    -s 1000 \
    -v \
    --threads 16
# Takes about 25 min for 50,000 genomes
# Size = 94 MB
```

```python
%%bash
### Create an inverted sketch for preclustering
sketchlib inverted build \
    -f fasta_files.tsv \
    --species-names /mmfs1/gscratch/pedslabs_hoffman/carsonjm/CFPhageome/repos/uhvdb-manuscript/host_genome_id/combined_inverted_clusters.tsv \
    -o chunk_102 \
    --write-skq \
    -s 10 \
    -k 21 \
    -v \
    --threads 16
### Takes about 8min for 50,000 genomes
# size = 1.8 MB
```

```python
%%bash
### Calculate distance within sketch
sketchlib inverted precluster \
    chunk_102.ski \
    --skd chunk_102 \
    -o chunk_102_precluster \
    --ani \
    --threads 16 \
    --verbose \
    --knn 100
# Takes 8s for 50,000 genomes
# Size = 382 MB
```

```python
%%bash
### Calculate distance within sketch
sketchlib dist \
    chunk_102_1k.skd \
    -o chunk_102_1k_dist \
    -k 21 \
    --ani \
    --threads 16 \
    --verbose \
    --knn 100
# Takes 53 s for 50,000 genomes
# Size = 422 MB  
```
